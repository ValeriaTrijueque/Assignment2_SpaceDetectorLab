```python
def parse_blev_file(filename, chip=0):
    with fits.open(filename) as hdul:
        data = hdul[4].data
        odd_vals = data[:,0::2]
        even_vals = data[:,1::2]
        return np.mean(odd_vals[odd_vals>0]), np.mean(even_vals[even_vals>0]), hdul.info()
filename = ('hst-data/Calibration/ubai2505m/ubai2505m_blevfile.fits')
odd, even, info = parse_blev_file(filename)
print(info)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from pathlib import Path
import numpy as np
from matplotlib.colors import rgb_to_hsv
import GEarth

here = Path('.')

white_images = sorted(list((here / 'sentinel-data/White_light').glob('*.jpg')))
for im in white_images:
    print(im)
    
```

```python
plt.close('all')
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(10, 12), tight_layout=True)
axes = axes.ravel()

titles = ["2014", "2016", "2020", "2021", "2022", "2023"]
for i, ax in enumerate(axes):
    img = mpimg.imread(white_images[i])
    ax.imshow(img)
    ax.set_title(titles[i])
    
```

```python
hsv_image2014, rgb_image2014 = GEarth.hsv_image(white_images[0])
hsv_image2016, rgb_image2016 = GEarth.hsv_image(white_images[1])
hsv_image2020, rgb_image2020 = GEarth.hsv_image(white_images[2])
hsv_image2021, rgb_image2021 = GEarth.hsv_image(white_images[3])
hsv_image2022, rgb_image2022 = GEarth.hsv_image(white_images[4])
hsv_image2023, rgb_image2023 = GEarth.hsv_image(white_images[5])

```

```python
plt.close('all')
fig, ((ax, ax1), (ax2, ax3), (ax4, ax5)) = plt.subplots(
    nrows=3, ncols=2, figsize=(10, 12), tight_layout=True
)

ax.imshow(hsv_image2014[:, :, 0])
ax.set_title("2014")

ax1.imshow(hsv_image2016[:, :, 0])
ax1.set_title("2016")

ax2.imshow(hsv_image2020[:, :, 0])
ax2.set_title("2020")

ax3.imshow(hsv_image2021[:, :, 0])
ax3.set_title("2021")

ax4.imshow(hsv_image2022[:, :, 0])
ax4.set_title("2022")

ax5.imshow(hsv_image2023[:, :, 0])
ax5.set_title("2023")


plt.show()
```

```python
wl = mpimg.imread(white_images[0])
fig, ax = plt.subplots(figsize=(9, 4))
ax.imshow(wl)
ax.set_xlim(2000,2400)
ax.set_ylim(2000, 1950)
plt.show()
```

```python
pixels = 2375 - 2062
size = 600 / pixels
area_pixel = size*size
area_pixel
```

```python
land_area_2014 = GEarth.calculate_land_area(hsv_image2014[:, :, 0], area_pixel)
land_area_2016 = GEarth.calculate_land_area(hsv_image2016[:, :, 0], area_pixel)
land_area_2020 = GEarth.calculate_land_area(hsv_image2020[:, :, 0], area_pixel)
land_area_2021 = GEarth.calculate_land_area(hsv_image2021[:, :, 0], area_pixel)
land_area_2022 = GEarth.calculate_land_area(hsv_image2022[:, :, 0], area_pixel)
land_area_2023 = GEarth.calculate_land_area(hsv_image2023[:, :, 0], area_pixel)
print(f"Land area of Hunga Tonga in 2014: {land_area_2014}m^2" )
print(f"Land area of Hunga Tonga in 2016: {land_area_2016}m^2" )
print(f"Land area of Hunga Tonga in 2020: {land_area_2020}m^2" )
print(f"Land area of Hunga Tonga in 2021: {land_area_2021}m^2" )
print(f"Land area of Hunga Tonga in 2022: {land_area_2022}m^2" )
print(f"Land area of Hunga Tonga in 2023: {land_area_2023}m^2" )
```

```python

fig,ax = plt.subplots(figsize=(9,7))
xpoints = np.array(["2014", "2016", "2020","2021","2022","2023"])
ypoints = np.array([land_area_2014, land_area_2016, land_area_2020,land_area_2021,land_area_2022,land_area_2023])

plt.scatter(xpoints, ypoints, color = 'blue')
plt.plot(xpoints, ypoints, color= 'blue', linestyle='--')
plt.xlabel("Year")
plt.ylabel("Land area in m^2")
plt.grid()
plt.show()
```

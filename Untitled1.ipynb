```python
file = ('hst-data/Calibration/ubai2505m/ubai2505m_maskfile.fits')
with fits.open(file) as hdul:
        header_mask= hdul[0].header
        data_mask = hdul[0].data
        info = hdul.info()
data_mask
```

```python
import numpy as np
masked_raw = np.where(data_mask == 0, data_raw, np.nan)
masked_raw
```

```python
import numpy as np
def parse_blev_file(filename, chip=1):
    with fits.open(filename) as hdul:


        data = hdul[chip].data
        odd = data_mask[:,0::2]
        even = data_mask[:,1::2]
            
        return np.mean(odd[odd>0]), np.mean(even[even>0]), hdul.info()
filename = ('hst-data/Calibration/ubai2505m/ubai2505m_blevfile.fits')
odd, even, info = parse_blev_file(filename)
odd
```

```python
masked_raw[:,  0::2] -= int(odd)
masked_raw[:, 1::2] -= int(even)
```
